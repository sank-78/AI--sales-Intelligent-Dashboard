```python
import os
import pandas as pd
from ucimlrepo import fetch_ucirepo

CSV_FILENAME = "online_retail.csv"

# 1. FORCE THE DELETION OF THE INCOMPLETE/BROKEN LOCAL FILE
if os.path.exists(CSV_FILENAME):
    print(f"⚠️ Found an incomplete local file. Checking columns...")
    test_df = pd.read_csv(CSV_FILENAME, nrows=5)
    if 'InvoiceNo' not in test_df.columns:
        print("🗑️ 'InvoiceNo' is missing! Deleting the incomplete CSV file to re-download...")
        os.remove(CSV_FILENAME)

# 2. DOWNLOAD EVERYTHING (INCLUDING HIDDEN ID COLUMNS) FROM UCI
if not os.path.exists(CSV_FILENAME):
    print("🛰️ Fetching complete dataset from UCI ML Repository (ID 352)...")
    try:
        # Fetch the entire package from the repository
        online_retail = fetch_ucirepo(id=352)
        
        # Extract all parts of the dataset structure
        X = online_retail.data.features   # Description, Quantity, InvoiceDate, UnitPrice, CustomerID, Country
        y = online_retail.data.targets    # Any target targets if applicable
        ids = online_retail.data.ids      # THIS IS WHERE INVOICENO AND STOCKCODE ARE HIDDEN! [1]
        
        # Create a list of parts that actually have data
        data_parts = []
        if ids is not None and not ids.empty:
            data_parts.append(ids)
        if X is not None and not X.empty:
            data_parts.append(X)
        if y is not None and not y.empty:
            data_parts.append(y)
            
        # Stitch all columns together side-by-side
        df = pd.concat(data_parts, axis=1)
        
        # Clean column spaces right away
        df.columns = df.columns.str.strip()
        
        # Save the complete file locally
        df.to_csv(CSV_FILENAME, index=False)
        print(f"✅ Success! Complete data saved as '{CSV_FILENAME}'")
        
    except Exception as e:
        print(f"❌ Download failed: {e}. Check your internet connection.")
        exit()

# 3. LOAD COMPLETE FILE OFF-LINE WITH FIXED DATA TYPES
print(f"📦 Loading complete '{CSV_FILENAME}' into pipeline...")
df = pd.read_csv(CSV_FILENAME, dtype={'InvoiceNo': str, 'StockCode': str})
df.columns = df.columns.str.strip()

print("🔍 Confirmed Available Columns now:", list(df.columns))

# 4. RUN YOUR CLEANING AND CALCULATIONS PIPELINE
try:
    # Ensure types are correct for computation
    df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
    df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors='coerce')
    df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'], errors='coerce')
    
    # Calculate Revenue
    df['Revenue'] = df['Quantity'] * df['UnitPrice']
    
    # Safely identify transaction cancellations
    df['IsCancellation'] = df['InvoiceNo'].str.upper().str.startswith('C', na=False)
    
    # Filter for active valid sales records
    sales = df[
        (df['IsCancellation'] == False) & 
        (df['Quantity'] > 0) & 
        (df['UnitPrice'] > 0) & 
        (df['InvoiceDate'].notna())
    ].copy()
    
    # Extract Month feature safely
    sales['Month'] = sales['InvoiceDate'].dt.to_period('M').astype(str)
    
    print("\n🚀 === PIPELINE COMPLETE ===")
    print(f"📊 Total Rows Recovered: {len(df):,}")
    print(f"✅ Cleaned Sales Rows:   {len(sales):,}")
    print(f"💰 Actual Total Revenue: ${sales['Revenue'].sum():,.2f}")

except KeyError as e:
    print(f"\n❌ KeyError: The column {e} was still not found.")
except Exception as e:
    print(f"\n❌ Execution Error: {e}")
```

```python
from sklearn.ensemble import IsolationForest
iso=IsolationForest(contamination=.08,random_state=42)
monthly['Anomaly']=iso.fit_predict(monthly[['Revenue','Orders','Units']].fillna(0))
monthly['AnomalyLabel']=np.where(monthly.Anomaly==-1,'Anomaly','Normal')
display(monthly[monthly.AnomalyLabel=='Anomaly'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
basket=pd.crosstab(sales.InvoiceNo,sales.Description).clip(upper=1)
top=basket.sum().sort_values(ascending=False).head(100).index
basket=basket[top]
sim=cosine_similarity(basket.T)
sim=pd.DataFrame(sim,index=top,columns=top)
rows=[]
for item in top:
    for rec,score in sim[item].sort_values(ascending=False).iloc[1:6].items():
        rows.append([item,rec,score])
recommendations=pd.DataFrame(rows,columns=['Product','RecommendedProduct','Similarity'])
display(recommendations.head(20))
```
